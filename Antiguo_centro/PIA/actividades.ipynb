25. Aplicar descuento — parámetro por defecto

```python
def calcular_precio(precio, descuento=10):
    return precio * (1-(descuento/100))

print(calcular_precio(100))
print(calcular_precio(100, 25))
```

26. Calcular salario — varios parámetros por defecto

```python
def calcular_salario(horas, precio_hora=12, bonus=0):
    return horas * precio_hora + bonus

print(calcular_salario(40)) #Las horas es indicado, el precio hora es 12 euros y no tien bonus, el precio hora y bonus los coge por defecto.
print(calcular_salario(40, 15)) #Las horas es indicado al igual que el precio de la hora, pero el bonus no por lo que es toma el valor por defecto.
print(calcular_salario(40, 15, 100)) #Aqui los 3 parametros son definidos.
```

27. Precio de una entrada — combinación de conceptos:
Las reglas son:
- Menores de 12 años: descuento del 50 %.
- Personas de 65 años o más: descuento del 30 %.
- Resto: pagan el precio normal.
- Si además es socio, se aplica un 10 % adicional sobre el precio obtenido anteriormente.


```python
def precio_entrada(edad, precio_base=10, es_socio=False):
    descuento = 0
    if edad < 12:
        descuento += 0.5
    elif edad >= 65:
        descuento += 0.3
    elif es_socio:
        return (precio_base * (1-descuento)) * 0.9
    return precio_base * (1-descuento)

print(precio_entrada(8))
print(precio_entrada(70,12))
print(precio_entrada(30,12,True))
```
