Escriba un programa que permita crear una lista de palabras y que, a continuación de tres  opciones: 
• Contar: Me pide una cadena, y me dice cuantas veces aparece en la lista 
• Modificar: Me pide una cadena, y otra cadena a modificar, y modifica todas a las  apariciones de la primera por la segunda en la lista. 
• Eliminar: Me pide una cadena, y la elimina de la lista. 
• Mostrar: Muestra la lista de cadenas 
• Terminar

```python
res = []
while True:
    print("• Contar: Me pide una cadena, y me dice cuantas veces aparece en la lista\n","• Modificar: Me pide una cadena, y otra cadena a modificar, y modifica todas a las  apariciones de la primera por la segunda en la lista. \n","• Eliminar: Me pide una cadena, y la elimina de la lista. \n","• Mostrar: Muestra la lista de cadenas\n","• Terminar\n")
    act = input("Dime una acción: ").upper()
    match  act:
        case "CONTAR":
            i = 0
            palabra = input("Dame una palabra para contarla")
            if palabra in res:
                for part in res:
                    if part == palabra:
                        i += 1
            print(f"La palabra '{palabra}' aparece {i} veces")
        case "MODIFICAR":
            palabra_or = input("Dame una palabra para cambiarla")
            palabra_nw = input("Dame una palabra nueva")
            if palabra_or in res:
                for indice, part in enumerate(res):
                    if part == palabra_or:
                        res[indice] = palabra_nw
        case "ELIMINAR":
            palabra = input("Dame una palabra para eliminarla")
            if palabra in res:
                for indice, part in enumerate(res):
                    if part == palabra:
                        res = res[:indice] + res[indice+1:]
        case "MOSTRAR":
            print(res)
        case "TERMINAR":
            break
```

Actividad 19: Suma de elementos
Dada una lista de números, crea una función que recorra la lista y devuelva la suma de todos sus elementos.
Ejemplo: sumar([1, 2, 3, 4]) debería devolver 10.

```python
def sumar(lista):
    res = 0
    for num in lista:
        res += num
    return res
```

Actividad 23: Eliminar duplicados
Dada una lista de números, crea una función que recorra la lista y elimine los elementos duplicados, devolviendo una nueva lista sin repeticiones.
Ejemplo: eliminarDuplicados([3, 5, 3, 7, 5, 9]) debería devolver [3, 5, 7, 9].

```python
def eliminarDuplicados(lista):
    res = []
    for valor in lista:
        if valor not in res:
            res.append(valor)
    return res
```

25. Transponer una Matriz
Escribe un programa que, dada una matriz (lista de listas), devuelva su transpuesta.
python

```python
def traspuesta(matriz):
    res = []
    for x in range(len(matriz)):
        res.append([])
    for indx ,x in enumerate(matriz):
        for indy, y in enumerate(matriz[indx]):
            res[indx][indy] = matriz[indy][indx]
    return res
```

26. Concatenar Elementos de una Fila
Dada una lista de listas que contiene cadenas, concatena todas las cadenas de una fila específica.

```python
def concatenar(matriz,fila):
    res = ""
    for palabra in matriz[fila]:
        res += palabra
    return res
```

27. Buscar una Palabra en la Lista
Dada una lista de listas con cadenas, escribe un programa que busque una palabra específica y devuelva su posición (índice de la fila y la columna).

```python
def localizar(matriz,palabra):
    for indx ,x in enumerate(matriz):
        for indy, y in enumerate(matriz[indx]):
            if matriz[indx][indy] == palabra:
                res = [indx,indy]
                return res
```

28. Contar Elementos
Dada una lista de tuplas que representa pares de valores, cuenta cuántas veces aparece cada valor en las tuplas.

```python
def contar_el(lista,valor):
    res = 0
    for tupla in lista:
        for v in tupla:
            if v == valor:
                res += 1
    return res 
```
