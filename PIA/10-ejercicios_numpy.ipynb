## 1️⃣ Crear un array y extraer la diagonal
Crea un array 4×4 con los números del 0 al 15 y extrae la diagonal principal.

```python
import numpy as np

cuadrado = np.arange(0,16).reshape((4,4))
res = []
i = 0
for fila in cuadrado:
    res.append(fila[i])
    i += 1
    
print(cuadrado)
print(res)
```

## 4️⃣ Estandarizar columnas (z-score)

Crea una matriz aleatoria 5×4 con `np.random.randn`. Calcula la media y desviación por columna y estandariza.

```python
import numpy as np

# 1. Crear una matriz aleatoria de 5x4 usando np.random.randn
matriz = np.random.randn(5, 4)
print("Matriz Original:\n", matriz)

# 2. Calcular la media por columna (axis=0)
medias = np.mean(matriz, axis=0)
print("\nMedia por columna:\n", medias)

# 3. Calcular la desviación estándar por columna (axis=0)
desviaciones = np.std(matriz, axis=0)
print("\nDesviación estándar por columna:\n", desviaciones)

# 4. Estandarizar la matriz
matriz_estandarizada = (matriz - medias) / desviaciones
print("\nMatriz Estandarizada:\n", matriz_estandarizada)

# Verificación opcional: la media de la matriz estandarizada debe ser ~0 y su desv. estándar ~1 por columna
print("\n[Verificación] Nueva media por columna:\n", np.mean(matriz_estandarizada, axis=0))
print("[Verificación] Nueva desv. estándar por columna:\n", np.std(matriz_estandarizada, axis=0))
```

## 5️⃣ Borde a 1, interior a 0

Crea una matriz 6×6 con los números del 0 al 35 y pon el borde a 1 y el interior a 0.

```python
matriz = np.arange(0,36).reshape(6,6)
print(matriz)

# Modificar el interior (filas de la 1 a la 4, columnas de la 1 a la 4) para que sean 0
matriz[1:-1, 1:-1] = 0

# Modificar los bordes asignándoles el valor 1
matriz[0, :] = 1   # Fila superior
matriz[-1, :] = 1  # Fila inferior
matriz[:, 0] = 1   # Columnna izquierda
matriz[:, -1] = 1  # Columna derecha

print("\nMatriz modificada:")
print(matriz)
```

## 6️⃣ Matriz de distancias por broadcasting

Dado `x = np.array([0, 3, 5, 9])`, construye una matriz `D` donde `D[i,j] = |x[i] - x[j]|`.

```python
x = np.array([0,3,5,9])

D = np.abs(x[:, np.newaxis] - x)

print(x, "\n")
print(D)
```

## 7️⃣ Reordenar filas por la suma

Genera una matriz 4×5 con enteros aleatorios entre 0 y 9. Reordena las filas según la suma (descendente).

```python
import numpy as np

matriz = np.random.randint(0, 10, size=(4, 5))
print("Matriz original:")
print(matriz)

# 2. Calcular la suma de cada fila (eje 1)
sumas = matriz.sum(axis=1)
print("\nSumas de cada fila:", sumas)

# 3. Obtener los índices para ordenar las sumas de forma descendente
# np.argsort ordena de menor a mayor; al pasarle '-sumas' lo invierte
indices_descendentes = np.argsort(-sumas)

# 4. Reordenar las filas de la matriz usando indexación avanzada
matriz_ordenada = matriz[indices_descendentes]

print("\nMatriz ordenada (por suma descendente):")
print(matriz_ordenada)
```
