Actividad 6: Validación de entrada numérica simple
Descripción: Crear una función que pida al usuario que ingrese un número y que maneje la
excepción si el usuario introduce un valor no numérico. mostrando un mensaje y termine.
Objetivo: Introducir el concepto de manejo básico de excepciones con try....

```python
def pedir_numero():
    try:
        numero = int(input("Dame un número:"))
    except ValueError:
        print("No es un número")

pedir_numero()
```

Actividad 8: Usar un valor por defecto al fallar
Descripción: Pedir al usuario que ingrese la edad. Si no es un número . Si el valor no es
numérico, utiliza un valor por defecto 18.
Objetivo: Introducir el concepto de continuar la ejecución con un valor predeterminado.

```python
def pedir_edad():
    try:
        numero = int(input("Dame un número:"))
        return numero
    except ValueError:
        return 18

print(pedir_edad())
```

Actividad 9: Volver a pedir el dato hasta que sea correcto
Descripción: Crear una función que solicite al usuario su nombre. Si introduce una cadena
vacía o un número, vuelve a pedir el dato hasta que sea válido.

```python
def pedir_nombre():
    while True:
        try:
            nombre = input("Dime tu nombre")
            if nombre == "":
                raise ValueError("Cadena vacia")
            elif nombre.isnumeric():
                raise ValueError("El valor es un número")
            else:
                return nombre
        except ValueError as error:
            print(error)

pedir_nombre()
```

Actividad 10: Reintentar una operación varias veces
Descripción: Simular la lectura de un archivo (o alguna operación que podría fallar)
y reintentar hasta 3 veces si falla.
Objetivo: Introducir el uso de reintentos con excepciones en operaciones que pueden fallar.

```python
def leer_dirección():
    res = 0
    while 3 > res:
        try:
            dire = input("Dime tu dirreción")
            if dire == "":
                raise ValueError("Dirección Vacia")
            return dire
        except ValueError as error:
            res +=1
            print(error)

leer_dirección()
```

Actividad 11 adaptar no dividir por cero de la actividad funciones 13 a que recoja la
excepción y vuelva a pedir.

```python
def dividir():
    try:
        numero1 = int(input("Que numero quieres dividir? "))
        numero2 = int(input("Por cual numero quieres dividir? "))
        return numero1 / numero2
    except ZeroDivisionError:
        print("No se puede dividir por cero\n")
        dividir()

dividir()
```
