## 📝 Ejercicio 5: Diccionarios y listas

```python
# Crea una lista con 3 diccionarios que representen alumnos (nombre y nota)
# Recorre la lista e imprime el nombre del alumno y si está aprobado (nota >= 5)

# Tu código aquí
diccionarios = [{"Juan" : 5},{"Sara" : 3},{"Fran":8}]

for diccionario in diccionarios:
    for alumno, nota in diccionario.items():
        if nota >= 5:
            print(alumno)
```
