### Ejercicio 1: Import
Import pandas, numpy

```python
# Pista: import pandas as pd; import numpy as np
import pandas as pd
import numpy as np
```

```python
assert 'pd' in dir()
print('✓ OK')
```

### Ejercicio 2: Load data
https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv

```python
# Pista: pd.read_csv()
baby_names = pd.read_csv('https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv')
baby_names.shape
```

```python
assert baby_names.shape[0] > 0
print('✓ Cargado')
```

### Ejercicio 3: Preview
First rows

```python
# Pista: .head()
baby_names.head()
```

```python
print('OK')
print('✓ OK')
```

### Ejercicio 4: Drop cols
Delete 'Unnamed: 0' and 'Id'

```python
# Pista: .drop(['Unnamed: 0', 'Id'], axis=1)
bn_clean = baby_names.drop(['Unnamed: 0', 'Id'], axis=1)
bn_clean.head()
```

```python
assert bn_clean is not None
print('✓ Limpio')
```

### Ejercicio 5: Gender count
More male or female names?

```python
# Pista: .groupby('Gender').size()
gender_count = bn_clean.groupby('Gender').size()
print(gender_count)
print('Más registros de:', gender_count.idxmax())
```

```python
assert gender_count is not None
print('✓ OK')
```

### Ejercicio 6: Group by name
Groupby name

```python
# Pista: .groupby('Name')
by_name = bn_clean.groupby('Name')
by_name.Count.sum().head()
```

```python
assert by_name is not None
print('✓ OK')
```

### Ejercicio 7: Unique names
How many different names?

```python
# Pista: .nunique() or .groupby().ngroups
num_names = bn_clean.Name.nunique()
print(num_names)
```

```python
assert num_names > 0
print('✓ OK')
```

### Ejercicio 8: Most common
Name with most occurrences

```python
# Pista: .groupby('Name').Count.sum().idxmax()
most_common = bn_clean.groupby('Name').Count.sum().idxmax()
print(most_common, bn_clean.groupby('Name').Count.sum().max())
```

```python
assert most_common is not None
print('✓ OK')
```

### Ejercicio 9: Least common
How many with least?

```python
# Pista: ...min()...
name_totals = bn_clean.groupby('Name').Count.sum()
least_count = name_totals.min()
print('Ocurrencias mínimas:', least_count)
print('Nombres con ese mínimo:', (name_totals == least_count).sum())
```

```python
assert least_count > 0
print('✓ OK')
```

### Ejercicio 10: Median
Median name occurrence

```python
# Pista: .groupby('Name').Count.sum().median()
med = bn_clean.groupby('Name').Count.sum().median()
print(med)
```

```python
assert med > 0
print('✓ OK')
```

### Ejercicio 11: Std
Standard deviation

```python
# Pista: .std()
std_val = bn_clean.groupby('Name').Count.sum().std()
print(std_val)
```

```python
assert std_val > 0
print('✓ OK')
```

### Ejercicio 12: Describe
Summary stats

```python
# Pista: .describe()
summary = bn_clean.groupby('Name').Count.sum().describe()
summary
```

```python
assert summary is not None
print('✓ OK')
```

### Ejercicio 13: Top 10
Top 10 names

```python
# Pista: .groupby('Name').Count.sum().nlargest(10)
top10 = bn_clean.groupby('Name').Count.sum().nlargest(10)
top10
```

```python
assert len(top10) == 10
print('✓ OK')
```

### Ejercicio 14: Percentiles
25th, 50th, 75th

```python
# Pista: .quantile([.25, .5, .75])
perc = bn_clean.groupby('Name').Count.sum().quantile([.25, .5, .75])
perc
```

```python
assert perc is not None
print('✓ OK')
```

### Ejercicio 15: Trend
Names trend over time

```python
# Pista: .groupby(['Year', 'Name']).Count.sum()
trend = bn_clean.groupby(['Year', 'Name']).Count.sum()
trend.head(10)
```

```python
assert trend is not None
print('✓ OK')
```
