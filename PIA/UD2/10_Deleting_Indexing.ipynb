# Deleting & Indexing

### Dataset: Iris

### Ejercicio 1: Import
Import pandas, numpy

```python
# Pista: import pandas as pd
import pandas as pd
import numpy as np
```

```python
assert 'pd' in dir()
print('✓ OK')
```

### Ejercicio 2: Load iris
https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data

```python
# Pista: pd.read_csv()
iris = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header=None)
iris.head()
```

```python
assert iris.shape[0] > 0
print('✓ Cargado')
```

### Ejercicio 3: Add columns
sepal_length, sepal_width, petal_length, petal_width, class

```python
# Pista: iris.columns = [...]
iris_cols = iris.copy()
iris_cols.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
iris_cols.head()
```

```python
assert iris_cols.shape[1] == 5
print('✓ OK')
```

### Ejercicio 4: Check nulls
Missing values?

```python
# Pista: .isnull().sum()
nulls = iris_cols.isnull().sum()
print(nulls)
```

```python
assert nulls is not None
print('✓ OK')
```

### Ejercicio 5: Set NaN
petal_length rows 10-29 to NaN

```python
# Pista: iris.loc[10:29, 'petal_length'] = np.nan
iris_nan = iris_cols.copy()
iris_nan.loc[10:29, 'petal_length'] = np.nan
print(iris_nan['petal_length'].isnull().sum(), 'NaN en petal_length')
```

```python
assert iris_nan.isnull().sum().sum() > 0
print('✓ OK')
```

### Ejercicio 6: Fill NaN
Fill NaN with 1.0

```python
# Pista: .fillna(1.0)
iris_fill = iris_nan.fillna(1.0)
iris_fill.loc[8:31, 'petal_length']
```

```python
assert iris_fill.isnull().sum().sum() == 0
print('✓ OK')
```

### Ejercicio 7: Drop class
Delete class column

```python
# Pista: .drop('class', axis=1)
iris_no_class = iris_fill.drop('class', axis=1)
iris_no_class.head()
```

```python
assert 'class' not in iris_no_class.columns
print('✓ OK')
```

### Ejercicio 8: First 3 NaN
Set first 3 rows to NaN

```python
# Pista: iris.loc[0:2] = np.nan
iris_head_nan = iris_no_class.copy()
iris_head_nan.loc[0:2] = np.nan
iris_head_nan.head()
```

```python
assert iris_head_nan.isnull().any().any()
print('✓ OK')
```

### Ejercicio 9: Dropna
Delete rows with NaN

```python
# Pista: .dropna()
iris_clean = iris_head_nan.dropna()
print(iris_head_nan.shape, '->', iris_clean.shape)
```

```python
assert iris_clean.isnull().sum().sum() == 0
print('✓ OK')
```

### Ejercicio 10: Reset index
Reset index to 0,1,2,...

```python
# Pista: .reset_index(drop=True)
iris_reset = iris_clean.reset_index(drop=True)
iris_reset.head()
```

```python
assert iris_reset.index[0] == 0
print('✓ OK')
```

### Ejercicio 11: Set index
Set column as index

```python
# Pista: .set_index('sepal_length')
iris_indexed = iris_reset.set_index('sepal_length')
iris_indexed.head()
```

```python
assert iris_indexed.index.name == 'sepal_length'
print('✓ OK')
```

### Ejercicio 12: Drop duplicates
Remove duplicate rows

```python
# Pista: .drop_duplicates()
iris_unique = iris_indexed.drop_duplicates()
print(iris_indexed.shape, '->', iris_unique.shape)
```

```python
assert iris_unique is not None
print('✓ OK')
```

### Ejercicio 13: Filter index
Select by index value

```python
# Pista: .loc[5.0]
by_index = iris_unique.loc[5.0]
by_index
```

```python
assert by_index is not None
print('✓ OK')
```

### Ejercicio 14: Multi index
Create MultiIndex

```python
# Pista: pd.MultiIndex.from_tuples([...])
multi_idx = pd.MultiIndex.from_tuples(
    [('setosa', 1), ('setosa', 2), ('versicolor', 1), ('versicolor', 2), ('virginica', 1), ('virginica', 2)],
    names=['species', 'sample'])
multi_idx
```

```python
assert multi_idx is not None
print('✓ OK')
```

### Ejercicio 15: Index operations
.index.values, .index.name

```python
# Pista: .index
print(iris_indexed.index.values[:10])
print(iris_indexed.index.name)
```
