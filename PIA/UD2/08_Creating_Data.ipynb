# Creating Series and DataFrames

### Dataset: Pokemon

### Ejercicio 1: Import
Import pandas

```python
# Pista: import pandas as pd
import pandas as pd
```

```python
assert 'pd' in dir()
print('✓ OK')
```

### Ejercicio 2: Create dict
Create Pokemon dictionary

```python
# Pista: data = {'Name': [...], 'Type': [...], ...}
pokemon_dict = {
    'Name': ['Bulbasaur', 'Charmander', 'Squirtle', 'Pikachu'],
    'Pokedex': [1, 4, 7, 25],
    'Type': ['Grass', 'Fire', 'Water', 'Electric'],
    'HP': [45, 39, 44, 35],
    'Evolution': ['Ivysaur', 'Charmeleon', 'Wartortle', 'Raichu'],
}
```

```python
assert isinstance(pokemon_dict, dict)
print('✓ OK')
```

### Ejercicio 3: Create DF
DataFrame from dict

```python
# Pista: pokemon = pd.DataFrame(pokemon_dict)
pokemon = pd.DataFrame(pokemon_dict)
pokemon
```

```python
assert isinstance(pokemon, pd.DataFrame)
print('✓ OK')
```

### Ejercicio 4: Reorder cols
Reorder to: Name, Type, HP, Evolution, Pokedex

```python
# Pista: pokemon[['Name', 'Type', 'HP', 'Evolution', 'Pokedex']]
pokemon_ordered = pokemon[['Name', 'Type', 'HP', 'Evolution', 'Pokedex']]
pokemon_ordered
```

```python
assert list(pokemon_ordered.columns)[0] == 'Name'
print('✓ OK')
```

### Ejercicio 5: Add column
Add 'Place' column

```python
# Pista: pokemon['Place'] = ['Kanto', 'Kanto', ...]
with_place = pokemon_ordered.copy()
with_place['Place'] = ['Kanto', 'Kanto', 'Kanto', 'Kanto']
with_place
```

```python
assert 'Place' in with_place.columns
print('✓ OK')
```

### Ejercicio 7: Create Series
Create a Series

```python
# Pista: pd.Series([1, 2, 3])
series = pd.Series([1, 2, 3])
series
```

```python
assert isinstance(series, pd.Series)
print('✓ OK')
```

### Ejercicio 8: Series with index
Series with custom index

```python
# Pista: pd.Series([...], index=['a', 'b', 'c'])
indexed = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
indexed
```

```python
assert len(indexed.index) == 3
print('✓ OK')
```

### Ejercicio 9: Concat series
Concatenate Series

```python
# Pista: pd.concat([s1, s2])
s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])
concat_s = pd.concat([s1, s2], ignore_index=True)
concat_s
```

```python
assert concat_s is not None
print('✓ OK')
```

### Ejercicio 10: From lists
DF from lists

```python
# Pista: pd.DataFrame({'col1': [...], 'col2': [...]})
df_lists = pd.DataFrame({'col1': [1, 2, 3], 'col2': ['a', 'b', 'c']})
df_lists
```

```python
assert df_lists.shape[1] == 2
print('✓ OK')
```
