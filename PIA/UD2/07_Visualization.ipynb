# Visualization

### Dataset: Titanic

### Ejercicio 1: Import
Import pandas, matplotlib, seaborn

```python
# Pista: import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
assert 'pd' in dir()
print('✓ OK')
```

### Ejercicio 2: Load titanic
https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Disaster/train.csv

```python
# Pista: pd.read_csv()
titanic = pd.read_csv('https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Disaster/train.csv')
titanic.head()
```

```python
assert titanic.shape[0] > 0
print('✓ Cargado')
```

### Ejercicio 3: Set index
PassengerId as index

```python
# Pista: .set_index('PassengerId')
titanic_idx = titanic.set_index('PassengerId')
titanic_idx.head()
```

```python
assert 'PassengerId' in titanic_idx.index.name or titanic_idx.index.name is None
print('✓ OK')
```

### Ejercicio 4: Pie gender
Gender distribution pie

```python
# Pista: titanic['Sex'].value_counts().plot(kind='pie')
plt.figure()
pie = titanic['Sex'].value_counts().plot(kind='pie', autopct='%.1f%%', title='Distribución por género')
pie.set_ylabel('')
plt.show()
```

```python
assert pie is not None
print('✓ OK')
```

### Ejercicio 5: Scatter
Fare vs Age by gender

```python
# Pista: .plot(x='Age', y='Fare', kind='scatter', c='Sex')
colors = titanic['Sex'].map({'male': 'tab:blue', 'female': 'tab:red'})
scatter = titanic.plot(x='Age', y='Fare', kind='scatter', c=colors, title='Fare vs Age (azul=male, rojo=female)')
plt.show()
```

```python
assert scatter is not None
print('✓ OK')
```

### Ejercicio 6: Survived count
How many survived?

```python
# Pista: titanic['Survived'].sum()
survived = titanic['Survived'].sum()
print(survived)
```

```python
assert survived > 0
print('✓ OK')
```

### Ejercicio 7: Histogram fare
Fare distribution

```python
# Pista: titanic['Fare'].plot(kind='hist')
plt.figure()
hist = titanic['Fare'].plot(kind='hist', bins=30, title='Distribución de Fare')
hist.set_xlabel('Fare')
plt.show()
```

```python
assert hist is not None
print('✓ OK')
```

### Ejercicio 8: Bar plot
Bar chart of classes

```python
# Pista: titanic['Pclass'].value_counts().plot(kind='bar')
plt.figure()
bar = titanic['Pclass'].value_counts().sort_index().plot(kind='bar', title='Pasajeros por clase')
bar.set_xlabel('Pclass')
plt.show()
```

```python
assert bar is not None
print('✓ OK')
```

### Ejercicio 9: Boxplot age
Age boxplot

```python
# Pista: titanic['Age'].plot(kind='box')
plt.figure()
box = titanic['Age'].plot(kind='box', title='Boxplot de Age')
plt.show()
```

```python
assert box is not None
print('✓ OK')
```

### Ejercicio 10: Group viz
Survived by gender

```python
# Pista: titanic.groupby('Sex')['Survived'].mean()
surv_gender = titanic.groupby('Sex')['Survived'].mean()
print(surv_gender)
plt.figure()
surv_gender.plot(kind='bar', title='Tasa de supervivencia por género')
plt.show()
```

```python
assert surv_gender is not None
print('✓ OK')
```

### Ejercicio 11: Heatmap
Correlation heatmap

```python
# Pista: sns.heatmap(titanic.corr())
plt.figure(figsize=(8, 6))
heat = sns.heatmap(titanic.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```

```python
assert heat is not None
print('✓ OK')
```

### Ejercicio 12: Kde plot
Age KDE

```python
# Pista: titanic['Age'].plot(kind='kde')
plt.figure()
kde = titanic['Age'].plot(kind='kde', title='KDE de Age')
plt.show()
```

```python
assert kde is not None
print('✓ OK')
```

### Ejercicio 13: Multi plot
Age by Pclass

```python
# Pista: titanic.boxplot(column='Age', by='Pclass')
multi = titanic.boxplot(column='Age', by='Pclass')
plt.show()
```

```python
assert multi is not None
print('✓ OK')
```

### Ejercicio 14: Violin
Violin plot

```python
# Pista: sns.violinplot(data=titanic, x='Sex', y='Age')
plt.figure()
violin = sns.violinplot(data=titanic, x='Sex', y='Age')
plt.show()
```

```python
assert violin is not None
print('✓ OK')
```
