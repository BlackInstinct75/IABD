# Merge

### Dataset: Auto MPG

### Ejercicio 1: Import
Import pandas, numpy

```python
import pandas as pd
import numpy as np
```

```python
assert 'pd' in dir()
print('✓ OK')
```

### Ejercicio 2: Load cars1
https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv

```python
# Pista: pd.read_csv()
url1 = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv'
cars1 = pd.read_csv(url1)
cars1.head()
```

```python
assert cars1 is not None
print('✓ Cargado')
```

### Ejercicio 3: Load cars2
https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv

```python
# Pista: pd.read_csv()
url2 = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv'
cars2 = pd.read_csv(url2)
cars2.head()
```

```python
assert cars2 is not None
print('✓ Cargado')
```

### Ejercicio 4: Clean cars1
Remove unnamed columns

```python
# Pista: .loc[:, 'mpg':'car']
cars1_clean = cars1.loc[:, 'mpg':'car']
cars1_clean.head()
```

```python
assert cars1_clean is not None
print('✓ Limpio')
```

### Ejercicio 5: Check shape
Shape of each df

```python
# Pista: .shape
print('cars1 original:', cars1.shape)
print('cars1_clean:   ', cars1_clean.shape)
print('cars2:         ', cars2.shape)
```

```python
print(f'{cars1_clean.shape}, {cars2.shape}')
print('✓ Revisado')
```

### Ejercicio 6: Concat
Join cars1 and cars2

```python
# Pista: pd.concat([cars1, cars2])
# ignore_index=True evita índices duplicados
cars = pd.concat([cars1_clean, cars2], ignore_index=True)
print(cars.shape)
```

```python
assert cars is not None
print('✓ Concatenado')
```

### Ejercicio 7: Create owners
Random integers 15000-73000

```python
# Pista: np.random.randint(15000, 73001, size=len(cars))
owners = np.random.randint(15000, 73001, size=len(cars))
owners[:5]
```

```python
assert len(owners) == len(cars)
print('✓ Creado')
```

### Ejercicio 8: Add column
Add owners to cars

```python
# Pista: cars['owners'] = owners
cars_owners = cars.copy()
cars_owners['owners'] = owners
cars_owners.head()
```

```python
assert 'owners' in cars_owners.columns
print('✓ Añadido')
```

### Ejercicio 9: Merge condition
pd.merge on index

```python
# Pista: pd.merge(cars1, cars2, left_index=True, right_index=True)
# Las columnas con el mismo nombre reciben automáticamente los sufijos _x y _y
merged = pd.merge(cars1_clean, cars2, left_index=True, right_index=True)
print(merged.shape)
merged.head()
```

```python
assert merged is not None
print('✓ OK')
```

### Ejercicio 10: Inner join
Inner merge

```python
# Pista: pd.merge(..., how='inner')
# Solo índices presentes en ambos (0-197)
inner = pd.merge(cars1_clean, cars2, left_index=True, right_index=True, how='inner')
print(inner.shape)
```

```python
assert inner is not None
print('✓ OK')
```

### Ejercicio 11: Left join
Left merge

```python
# Pista: pd.merge(..., how='left')
# Todos los índices de cars1_clean
left = pd.merge(cars1_clean, cars2, left_index=True, right_index=True, how='left')
print(left.shape)
```

```python
assert left is not None
print('✓ OK')
```

### Ejercicio 12: Outer join
Outer merge

```python
# Pista: pd.merge(..., how='outer')
# Unión de índices: cars2 tiene 2 filas más (198 y 199), que quedan con NaN en las columnas de cars1
outer = pd.merge(cars1_clean, cars2, left_index=True, right_index=True, how='outer')
print(outer.shape)
```

```python
assert outer is not None
print('✓ OK')
```

### Ejercicio 13: On column
Merge on specific column

```python
# Pista: pd.merge(..., on='key')
# No existe una columna 'key'; usamos 'car' (nombre del coche) como clave
on_col = pd.merge(cars1_clean, cars2, on='car')
print(on_col.shape)
on_col.head()
```

```python
assert on_col is not None
print('✓ OK')
```

### Ejercicio 14: Suffixes
Add suffixes for conflicts

```python
# Pista: pd.merge(..., suffixes=('_left', '_right'))
suff = pd.merge(cars1_clean, cars2, left_index=True, right_index=True,
                suffixes=('_left', '_right'))
print(suff.columns.tolist())
```

```python
assert suff is not None
print('✓ OK')
```

### Ejercicio 15: Append
Add rows with append/concat

```python
# Pista: pd.concat([cars1, cars2], ignore_index=True)
# DataFrame.append fue eliminado en pandas 2.0; concat es el equivalente
appended = pd.concat([cars1_clean, cars2], ignore_index=True)
print(appended.shape)
```

```python
assert len(appended) > 0
print('✓ OK')
```
