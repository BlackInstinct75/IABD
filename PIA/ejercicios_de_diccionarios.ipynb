Ejercicio 1
Escribe un programa python que pida un número por teclado y que cree un diccionario
cuyas claves sean desde el número 1 hasta el número indicado, y los valores sean los
cuadrados de las claves.

```python
numero = int(input("Dame un número: "))
res = {}

for num in range(1,numero+1):
    res[num] = num**2

print(res)
```

Ejercicio 2
Escribe un programa que lea una cadena y devuelva un diccionario con la cantidad de
apariciones de cada carácter en la cadena.

```python
cadena = input("Dame una cadena: ")

res = {}
for letra in cadena:
    aux = 0
    if letra not in res:
        for let in cadena:
            if letra == let:
                aux += 1
        res[letra] = aux
print(cadena)
print(res)
```

Ejercicio 3
Vamos a crear un programa en python donde vamos a declarar un diccionario para
guardar los precios de las distintas frutas. El programa pedirá el nombre de la fruta y la
cantidad que se ha vendido y nos mostrará el precio final de la fruta a partir de los datos
guardados en el diccionario. Si la fruta no existe nos dará un error. Tras cada consulta el
programa nos preguntará si queremos hacer otra consulta.

```python
frutas = {"Melon":12,"Sandia":20,"Pera":42}

while True:
    try:
        fru = input("Dime una fruta")
        print(f"La fruta {fru} se ha vendido {frutas[fru]}")
        op = input("Quieres hacer otra consulta? ")
        if op != "Si":
            break
    except KeyError:
        print("No tenemos esta fruta")
```

Ejercicio 4
Codifica un programa en python que nos permita guardar los nombres de los alumnos de
una clase y las notas que han obtenido. Cada alumno puede tener distinta cantidad de
notas. Guarda la información en un diccionario cuya claves serán los nombres de los
alumnos y los valores serán listas con las notas de cada alumno.
El programa pedirá el número de alumnos que vamos a introducir, pedirá su nombre e irá
pidiendo sus notas hasta que introduzcamos un número negativo. Al final el programa
nos mostrará la lista de alumnos y la nota media obtenida por cada uno de ellos. Nota: si
se introduce el nombre de un alumno que ya existe el programa nos dará un error.

```python
try:
    n_alu = int(input("Cuanto alumnos vas a introducir?"))

    res = {}
    for al in range(n_alu):
        nombre = input("Dime el nombre")
        if nombre in res:        
            raise ValueError()
        notas = []
        while True:
            nota = int(input("Dame la nota"))
            if nota < 0:
                break
            notas.append(nota)
        res[nombre] = notas
    print(res)
except ValueError:
    print("El nombre ya esta")
```

Ejercicio 5
Escribir un programa que implemente una agenda. En la agenda se podrán guardar
nombres y números de teléfono. El programa nos dará el siguiente menú:
• Añadir/modificar: Nos pide un nombre. Si el nombre se encuentra en la agenda,
debe mostrar el teléfono y, opcionalmente, permitir modificarlo si no es correcto.
Si el nombre no se encuentra, debe permitir ingresar el teléfono correspondiente.
• Buscar: Nos pide una cadena de caracteres, y nos muestras todos los contactos
cuyos nombres comiencen por dicha cadena.
• Borrar: Nos pide un nombre y si existe nos preguntará si queremos borrarlo de la
agenda.
• Listar: Nos muestra todos los contactos de la agenda.
Implementar el programa con un diccionario.

```python
res = {}
while True:
    print("• Añadir/modificar: Nos pide un nombre. Si el nombre se encuentra en la agenda,"
    "debe mostrar el teléfono y, opcionalmente, permitir modificarlo si no es correcto."
    "Si el nombre no se encuentra, debe permitir ingresar el teléfono correspondiente.\n"
    "• Buscar: Nos pide una cadena de caracteres, y nos muestras todos los contactos"
    "cuyos nombres comiencen por dicha cadena.\n"
    "• Borrar: Nos pide un nombre y si existe nos preguntará si queremos borrarlo de la"
    "agenda.\n"
    "• Listar: Nos muestra todos los contactos de la agenda.\n")

    opcion = input("Dime tu opción: ")
    match opcion:
        case "Añadir/modificar":
            nombre = input("Dime el nombre:")
            if nombre in res:
                print(res[nombre])
                op = input("Quieres cambiarlo? ")
                if op == "Si":
                    numero = input("Dime el número correcto: ")
                    res[nombre] = numero
            else:
                numero = input("Dime el número: ")
                res[nombre] = numero
        case "Buscar":
            nombre = input("Dime el nombre: ")
            print(res[nombre])
        case "Borrar":
            nombre = input("Dime el nombre: ")
            res.pop(nombre)
        case "Listar":
            print(res)
        case _:
            break
```
