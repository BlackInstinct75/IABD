### Ejercicio 1: Importar pandas
Import pandas

```python
# Pista: import pandas as pd
import pandas as pd
```

### Ejercicio 2: Cargar dataset
URL: https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/02_Filtering_%26_Sorting/Euro12/Euro_2012_stats_TEAM.csv

```python
# Pista: pd.read_csv()
euro12 = pd.read_csv('https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/02_Filtering_%26_Sorting/Euro12/Euro_2012_stats_TEAM.csv')
```

```python
assert euro12.shape[0] == 16
print('✓ 16 equipos')
```

### Ejercicio 3: Ver estructura
head() + info()

```python
# Pista: .head() e .info()
print(euro12.head())
euro12.info()
```

```python
print('OK')
print('✓ Revisado')
```

### Ejercicio 4: Columna Goals
Obtén Series de Goals

```python
# Pista: euro12['Goals']
goals = euro12['Goals']
```

```python
assert isinstance(goals, pd.Series)
print('✓ Series')
```

### Ejercicio 5: Contar equipos
¿Cuántos?

```python
# Pista: .shape[0]
num_equipos = euro12.shape[0]
```

```python
assert num_equipos == 16
print('✓ 16')
```

### Ejercicio 6: Columnas totales
¿Cuántas columnas?

```python
# Pista: .shape[1]
num_col = euro12.shape[1]
```

```python
assert num_col == 35
print('✓ 35')
```

### Ejercicio 7: Select 3 cols
Team, Yellow Cards, Red Cards

```python
# Pista: euro12[[...]]
discipline = euro12[['Team', 'Yellow Cards', 'Red Cards']]
```

```python
assert discipline.shape == (16, 3)
print('✓ OK')
```

### Ejercicio 8: Sort by cards
Red desc, Yellow desc

```python
# Pista: .sort_values()
sorted_d = discipline.sort_values(by=['Red Cards', 'Yellow Cards'], ascending=False)
```

```python
assert sorted_d is not None
print('✓ Ordenado')
```

### Ejercicio 9: Media amarillas
Mean of Yellow Cards

```python
# Pista: .mean()
media_y = euro12['Yellow Cards'].mean()
print(media_y)
```

```python
assert 7 < media_y < 8
print('✓ Media')
```

### Ejercicio 10: Filter >6 goals
Teams with >6 goals

```python
# Pista: euro12[euro12.Goals > 6]
teams_6 = euro12[euro12.Goals > 6]
```

```python
assert len(teams_6) > 0
print('✓ OK')
```

### Ejercicio 11: Teams start G
Team name starts with G

```python
# Pista: .str.startswith('G')
teams_g = euro12[euro12.Team.str.startswith('G')]
```

```python
assert len(teams_g) > 0
print('✓ OK')
```

### Ejercicio 12: First 7 cols
Select first 7

```python
# Pista: .iloc[:, 0:7]
first_7 = euro12.iloc[:, 0:7]
```

```python
assert first_7.shape[1] == 7
print('✓ OK')
```

### Ejercicio 13: No last 3
Exclude last 3

```python
# Pista: .iloc[:, :-3]
no_last = euro12.iloc[:, :-3]
```

```python
assert no_last.shape[1] == 32
print('✓ OK')
```

### Ejercicio 14: 3 teams accuracy
England, Italy, Russia

```python
# Pista: .loc[.isin(...)]
teams_p = euro12.loc[euro12.Team.isin(['England', 'Italy', 'Russia']), ['Team', 'Shooting Accuracy']]
```

```python
assert teams_p.shape[0] == 3
print('✓ OK')
```

### Ejercicio 15: Complex filter
>5 goals AND >80 passes

```python
# Pista: (cond1) & (cond2)
# Nota: el dataset no tiene 'Total passes'; la columna es 'Passes'
teams_f = euro12[(euro12.Goals > 5) & (euro12.Passes > 80)]
```

```python
assert len(teams_f) > 0
print('✓ OK')
```
